# 01 - Data Exploration

Load the merged dataset and reproduce the basic EDA stats used in the synopsis.
Keep this notebook thin: real logic lives in `src/`, this just calls it and plots.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/sepsis_merged.csv')
print(df.shape)
df.head()
```

```python
# class imbalance
print(df['SepsisLabel'].value_counts(normalize=True) * 100)
```

```python
# missingness per column
missing_pct = df.isna().mean().sort_values(ascending=False) * 100
missing_pct.head(15).plot(kind='barh', figsize=(8,6))
plt.xlabel('% missing')
plt.title('Worst missingness by column')
plt.tight_layout()
plt.savefig('../results/figures/missingness.png', dpi=150)
plt.show()
```
